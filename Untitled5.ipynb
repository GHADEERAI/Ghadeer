```python
import pandas as pd
from nltk.stem.isri import ISRIStemmer
from nltk.stem.snowball import SnowballStemmer
import re

# -----------------------------
# Text
# -----------------------------
text = "الحمد الله الرحمن الرحيم العالمين مالك الدين نعبد نستعين"
words = text.split()

# -----------------------------
# Simulated Khoja Stemmer
# -----------------------------
def khoja_stem(word):
    roots = {
        "الحمد": "حمد",
        "الله": "اله",
        "الرحمن": "رحم",
        "الرحيم": "رحم",
        "العالمين": "علم",
        "مالك": "ملك",
        "الدين": "دين",
        "نعبد": "عبد",
        "نستعين": "عون"
    }
    return roots.get(word, word)

# -----------------------------
# ISRI Stemmer
# -----------------------------
isri = ISRIStemmer()

# -----------------------------
# Light10-like Stemmer
# -----------------------------
def light10_stem(word):
    prefixes = ["وال", "بال", "كال", "فال", "لل", "ال", "و", "ف", "ب", "ك", "ل"]
    suffixes = ["كما", "هما", "كم", "كن", "نا", "ها", "هم", "هن",
                "ية", "ه", "ة", "ات", "ان", "ين", "ون", "ي", "ك", "ت", "ا"]
    for p in prefixes:
        if word.startswith(p) and len(word) > len(p) + 2:
            word = word[len(p):]
            break
    for s in suffixes:
        if word.endswith(s) and len(word) > len(s) + 2:
            word = word[:-len(s)]
            break
    return re.sub(r"[^\u0600-\u06FF]+", "", word)

# -----------------------------
# Snowball Stemmer
# -----------------------------
snowball = SnowballStemmer("arabic")

# -----------------------------
# Simulated Farasa Stemmer
# -----------------------------
def farasa_stem(word):
    roots = {
        "الحمد": "حمد",
        "الله": "اله",
        "الرحمن": "رحم",
        "الرحيم": "رحم",
        "العالمين": "علم",
        "مالك": "ملك",
        "الدين": "دين",
        "نعبد": "عبد",
        "نستعين": "عون"
    }
    return roots.get(word, word)

# -----------------------------
# Create DataFrame
# -----------------------------
data = {
    "Word": words,
    "Khoja": [khoja_stem(w) for w in words],
    "ISRI": [isri.stem(w) for w in words],
    "Light10": [light10_stem(w) for w in words],
    "Snowball": [snowball.stem(w) for w in words],
    "Farasa": [farasa_stem(w) for w in words]
}

df = pd.DataFrame(data)

df
```
